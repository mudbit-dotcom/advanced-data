```python
import pandas as pd
```

```python
df= pd.read_csv ("2025-2026 SNAP Data.csv")
df
```

```python
SNAP_BASE = df["Total SNAP Persons"].iat[7]
SNAP_BASE
```

```python
SNAP_BASE= df.iat [0,7]
```

```python
df["Total SNAP Persons"] = pd.to_numeric(df["Total SNAP Persons"].str.replace(",", ""), errors="coerce")
print(df["Total SNAP Persons"].dtype) 
```

```python
df["Total SNAP Persons"]= df["Total SNAP Persons"].apply(lambda x: (x - SNAP_BASE) / SNAP_BASE * 100)
print(df)
```
